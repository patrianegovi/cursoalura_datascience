```python
# visualizando com graficos

tmdb["original_language"].value_counts().index
```

```python
tmdb["original_language"].value_counts().values
```

```python
contagem_de_lingua = tmdb["original_language"].value_counts().to_frame().reset_index()
contagem_de_lingua.columns = ["original_language", "total"]
contagem_de_lingua.head()
```

```python
sns.barplot(data= contagem_de_lingua, x="original_language", y="total")
plt.title("Quantidade de filmes por lingua")
```

```python
contagem_de_lingua.plot(kind="pie", y="total", labels=contagem_de_lingua["original_language"])
# porque nao usar grafico de pizza
```

```python
total_por_lingua = tmdb["original_language"].value_counts()
total_geral = total_por_lingua.sum()
total_de_ingles = total_por_lingua.loc["en"]
total_do_resto = total_geral - total_de_ingles
print(total_geral, total_de_ingles, total_do_resto)
```

```python
dados = {
"linguas" : ["ingles", "outros"],
"total" : [total_de_ingles, total_do_resto]
}
dados = pd.DataFrame(dados)
dados
```

```python
sns.barplot(data=dados, x="linguas", y="total")
plt.title("Distribuição geral de linguas dos filmes")
```

```python
dados.plot(kind="pie", y="total", labels=dados["linguas"])
```

```python
# contar valores da coluna original language, transformar em um dataframe e reiniciar o indice da coluna.
# mudar nome da coluna para ter uma melhor interpretacao

contagem_de_lingua = tmdb["original_language"].value_counts().to_frame().reset_index()
contagem_de_lingua.columns = ["original_language", "total"]
contagem_de_lingua.head()
```

```python
# seaborn, original_language = x; total = y; contagem de lingua a fonte dos dados

sns.barplot(x='original_language', y='total', data=contagem_de_lingua)
```

```python
sns.countplot(data=tmdb, x="original_language")
```

```python
total_por_lingua = tmdb["original_language"].value_counts() # contagem de valores das linguas do DF.
total_geral = total_por_lingua.sum() #somar as linguas e salvar na variavel total geral
total_de_ingles = total_por_lingua.loc["en"] # subtrair do total geral os de lingua inglesa
total_do_resto = total_geral - total_de_ingles
print(total_geral, total_de_ingles, total_do_resto)
```

```python
# criando dicionario no pandas para gerar um dataframe.

dados = {
  "lingua" : ["ingles", "outros"],
  "total" : [total_de_ingles, total_do_resto]
}
dados = pd.DataFrame(dados)
dados
```

```python
sns.barplot(x="lingua", y="total", data = dados) # grafico em que x terá linguas e y as ocorrencias
```
