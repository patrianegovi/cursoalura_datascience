```python
filmes = pd.read_csv('https://raw.githubusercontent.com/alura-cursos/data-science-analise-exploratoria/main/Aula_0/ml-latest-small/movies.csv')
filmes.columns = ['filmeId', 'titulo', 'generos'] #alterando os nomes das colunas
filmes.head() #mostra as 5 primeitoas linhas
```

```python
notas.query("filmeId==1")["nota"].mean() #media das notas de Toy Story
```

```python
notas.query("filmeId==2")["nota"].mean() #media das notas de Jumanji
```

```python
medias_por_filme = notas.groupby("filmeId")["nota"].mean() #media de todos os filmes agrupados
medias_por_filme.head() #ver as 5 primeiras linhas
```

```python
medias_por_filme.plot(kind='hist') #histograma para ver a distribuicao das medias
```

```python
sns.boxplot(medias_por_filme) #boxplot para ver a distribuicao
```

```python
medias_por_filme.describe() #resumo dos dados
```

```python
# histrograma com Seaborn e Título

import matplotlib.pyplot as plt
sns.displot(medias_por_filme, kde=True)
plt.title("Histograma das médias dos filmes")
```
