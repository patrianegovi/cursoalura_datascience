```python
import pandas as pd
import numpy as np

# Criando dataframe de exemplo corrigido (ambas as semanas agora têm 5 elementos)
dados = np.array([[100, 120, 90, 130, 110], [80, 150, 70, 140, 60]])

df = pd.DataFrame(dados, index = ["Semana 1", "Semana 2"],
                  columns = ["Segunda", "Terca", "Quarta", "Quinta", "Sexta"])

# Calculando desvio padrão das visitas por semana (corrigido com espaços nos nomes)
desvio_padrao_semana1 = df.loc["Semana 1"].std()
desvio_padrao_semana2 = df.loc["Semana 2"].std()

print(df)
print("\nDesvio padrão da semana 1:", desvio_padrao_semana1)
print("Desvio padrão da semana 2:", desvio_padrao_semana2)
```
